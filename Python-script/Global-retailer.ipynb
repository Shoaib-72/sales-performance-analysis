```python
# Import required libraries

import pandas as pd
import numpy as np


from pathlib import Path

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.float_format', '{:.2f}'.format)

print("Libraries imported successfully!")
```

```python
# Load datasets

customers = pd.read_csv("Customers.csv",encoding="cp1252")
data_dictionary = pd.read_csv("Data_Dictionary.csv",encoding="cp1252")
exchange_rates = pd.read_csv("Exchange_Rates.csv",encoding="cp1252")
products = pd.read_csv("Products.csv",encoding="cp1252")
sales = pd.read_csv("Sales.csv",encoding="cp1252")
stores = pd.read_csv("Stores.csv",encoding="cp1252")

print("All datasets loaded successfully!")
```

```python
datasets = {
    "Customers": customers,
    "Data Dictionary": data_dictionary,
    "Exchange Rates": exchange_rates,
    "Products": products,
    "Sales": sales,
    "Stores": stores
}

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(f"DATASET: {name}")
    print("=" * 60)

    print("Shape:", df.shape)
    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:", df.duplicated().sum())

    print("\nFirst 5 Records:")
    display(df.head())
```

```python
# ==========================================
# A. CUSTOMERS DATA CLEANING
# ==========================================

customers_clean = customers.copy()

# 1. Remove completely empty rows and columns
customers_clean.dropna(how='all', inplace=True)
customers_clean.dropna(axis=1, how='all', inplace=True)

# 2. Remove duplicate records
customers_clean.drop_duplicates(inplace=True)

# 3. Remove unnecessary spaces from text columns
text_cols = customers_clean.select_dtypes(include='object').columns

for col in text_cols:
    customers_clean[col] = customers_clean[col].str.strip()

# 4. Convert CustomerKey into numeric format
customers_clean['CustomerKey'] = pd.to_numeric(
    customers_clean['CustomerKey'], errors='coerce'
)

# 5. Remove records with missing CustomerKey
customers_clean.dropna(subset=['CustomerKey'], inplace=True)

# 6. Remove duplicate CustomerKey records
customers_clean.drop_duplicates(
    subset=['CustomerKey'], keep='first', inplace=True
)

# 7. Convert Birthday into datetime
customers_clean['Birthday'] = pd.to_datetime(
    customers_clean['Birthday'], errors='coerce'
)

# 8. Standardize Gender
if 'Gender' in customers_clean.columns:
    customers_clean['Gender'] = (
        customers_clean['Gender'].str.title()
    )

# 9. Standardize Country and Continent
for col in ['Country', 'Continent', 'State']:
    if col in customers_clean.columns:
        customers_clean[col] = customers_clean[col].str.title()

# Final result
print("Customers Cleaning Completed!")
print("Shape:", customers_clean.shape)
print("Missing Values:\n", customers_clean.isnull().sum())
print("Duplicates:", customers_clean.duplicated().sum())

display(customers_clean.head())
```

```python
# ==========================================
# B. DATA DICTIONARY CLEANING
# ==========================================

dictionary_clean = data_dictionary.copy()

# 1. Remove completely empty rows and columns
dictionary_clean.dropna(how='all', inplace=True)
dictionary_clean.dropna(axis=1, how='all', inplace=True)

# 2. Remove exact duplicates
dictionary_clean.drop_duplicates(inplace=True)

# 3. Clean text columns
text_cols = dictionary_clean.select_dtypes(
    include='object'
).columns

for col in text_cols:
    dictionary_clean[col] = dictionary_clean[col].str.strip()

# 4. Standardize column names
dictionary_clean.columns = (
    dictionary_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
)

print("Data Dictionary Cleaning Completed!")
print("Shape:", dictionary_clean.shape)
print("Missing Values:\n", dictionary_clean.isnull().sum())

display(dictionary_clean.head())
```

```python
# ==========================================
# C. EXCHANGE RATES CLEANING
# ==========================================

exchange_clean = exchange_rates.copy()

# 1. Remove empty rows
exchange_clean.dropna(how='all', inplace=True)

# 2. Remove exact duplicates
exchange_clean.drop_duplicates(inplace=True)

# 3. Remove spaces from column names
exchange_clean.columns = exchange_clean.columns.str.strip()

# 4. Clean Currency column
exchange_clean['Currency'] = (
    exchange_clean['Currency'].str.strip().str.upper()
)

# 5. Convert Date into datetime
exchange_clean['Date'] = pd.to_datetime(
    exchange_clean['Date'], errors='coerce'
)

# 6. Convert Exchange into numeric
exchange_clean['Exchange'] = pd.to_numeric(
    exchange_clean['Exchange'], errors='coerce'
)

# 7. Remove records with missing essential fields
exchange_clean.dropna(
    subset=['Date', 'Currency', 'Exchange'], inplace=True
)

# 8. Remove zero or negative exchange rates
exchange_clean = exchange_clean[
    exchange_clean['Exchange'] > 0
]

# 9. Remove duplicate date and currency combinations
exchange_clean.drop_duplicates(
    subset=['Date', 'Currency'], keep='first', inplace=True
)

print("Exchange Rates Cleaning Completed!")
print("Shape:", exchange_clean.shape)
print("Missing Values:\n", exchange_clean.isnull().sum())

display(exchange_clean.head())
```

```python

# ==========================================
# D. PRODUCTS DATA CLEANING
# ==========================================

products_clean = products.copy()

# 1. Remove completely empty rows
products_clean.dropna(how='all', inplace=True)

# 2. Remove completely empty columns
products_clean.dropna(axis=1, how='all', inplace=True)

# 3. Remove exact duplicate records
products_clean.drop_duplicates(inplace=True)

# 4. Clean text columns
text_cols = products_clean.select_dtypes(
    include=['object', 'string']
).columns

for col in text_cols:
    products_clean[col] = products_clean[col].str.strip()

# 5. Convert ProductKey and category keys into numeric
key_columns = [
    'ProductKey',
    'SubcategoryKey',
    'CategoryKey'
]

for col in key_columns:
    if col in products_clean.columns:
        products_clean[col] = pd.to_numeric(
            products_clean[col], errors='coerce'
        )

# 6. Clean price columns before numeric conversion
price_columns = ['Unit Cost USD', 'Unit Price USD']

for col in price_columns:
    if col in products_clean.columns:

        products_clean[col] = (
            products_clean[col]
            .astype('string')
            .str.replace(r'[$,]', '', regex=True)
            .str.strip()
        )

        products_clean[col] = pd.to_numeric(
            products_clean[col], errors='coerce'
        )

# 7. Remove records without ProductKey
products_clean.dropna(
    subset=['ProductKey'], inplace=True
)

# 8. Remove duplicate ProductKey records
products_clean.drop_duplicates(
    subset=['ProductKey'],
    keep='first',
    inplace=True
)

# 9. Check and handle invalid negative prices
for col in price_columns:
    if col in products_clean.columns:
        products_clean.loc[
            products_clean[col] < 0, col
        ] = np.nan

# 10. Check missing values
print("Products Cleaning Completed!")

print("\nDataset Shape:", products_clean.shape)

print("\nMissing Values:")
print(products_clean.isnull().sum())

print("\nDuplicate Rows:",
      products_clean.duplicated().sum())

print("\nPrice Column Statistics:")
display(products_clean[price_columns].describe())

display(products_clean.head())
```

```python
# ==========================================
# E. SALES DATA CLEANING
# ==========================================

sales_clean = sales.copy()

# 1. Remove completely empty rows
sales_clean.dropna(how='all', inplace=True)

# 2. Remove exact duplicate records
sales_clean.drop_duplicates(inplace=True)

# 3. Clean text columns
text_cols = sales_clean.select_dtypes(
    include='object'
).columns

for col in text_cols:
    sales_clean[col] = sales_clean[col].str.strip()

# 4. Convert numeric columns
numeric_cols = [
    'Order Number',
    'Line Item',
    'CustomerKey',
    'StoreKey',
    'ProductKey',
    'Quantity'
]

for col in numeric_cols:
    if col in sales_clean.columns:
        sales_clean[col] = pd.to_numeric(
            sales_clean[col], errors='coerce'
        )

# 5. Convert Order Date into datetime
sales_clean['Order Date'] = pd.to_datetime(
    sales_clean['Order Date'], errors='coerce'
)

# 6. Standardize Currency Code
sales_clean['Currency Code'] = (
    sales_clean['Currency Code'].str.strip().str.upper()
)

# 7. Remove records missing essential transaction information
sales_clean.dropna(
    subset=[
        'Order Number',
        'Line Item',
        'Order Date',
        'CustomerKey',
        'ProductKey',
        'Quantity'
    ],
    inplace=True
)

# 8. Remove zero or negative quantities
sales_clean = sales_clean[
    sales_clean['Quantity'] > 0
]

# 9. Remove duplicate order line items
sales_clean.drop_duplicates(
    subset=['Order Number', 'Line Item'],
    keep='first',
    inplace=True
)

print("Sales Cleaning Completed!")
print("Shape:", sales_clean.shape)
print("Missing Values:\n", sales_clean.isnull().sum())

display(sales_clean.head())
```

```python
# ==========================================
# F. STORES DATA CLEANING
# ==========================================

stores_clean = stores.copy()

# 1. Remove completely empty rows
stores_clean.dropna(how='all', inplace=True)

# 2. Remove exact duplicates
stores_clean.drop_duplicates(inplace=True)

# 3. Clean text columns
text_cols = stores_clean.select_dtypes(
    include='object'
).columns

for col in text_cols:
    stores_clean[col] = stores_clean[col].str.strip()

# 4. Convert StoreKey into numeric
stores_clean['StoreKey'] = pd.to_numeric(
    stores_clean['StoreKey'], errors='coerce'
)

# 5. Convert Square Meters into numeric
stores_clean['Square Meters'] = pd.to_numeric(
    stores_clean['Square Meters'], errors='coerce'
)

# 6. Convert Open Date into datetime
stores_clean['Open Date'] = pd.to_datetime(
    stores_clean['Open Date'], errors='coerce'
)

# 7. Remove records without StoreKey
stores_clean.dropna(
    subset=['StoreKey'], inplace=True
)

# 8. Remove duplicate StoreKey
stores_clean.drop_duplicates(
    subset=['StoreKey'], keep='first', inplace=True
)

# 9. Remove invalid negative store area
stores_clean = stores_clean[
    stores_clean['Square Meters'].isna() |
    (stores_clean['Square Meters'] >= 0)
]

# 10. Standardize Country
if 'Country' in stores_clean.columns:
    stores_clean['Country'] = (
        stores_clean['Country'].str.title()
    )

print("Stores Cleaning Completed!")
print("Shape:", stores_clean.shape)
print("Missing Values:\n", stores_clean.isnull().sum())

display(stores_clean.head())
```

```python
# ==========================================
# SAVE ALL CLEANED DATASETS
# ==========================================

# Create output folder
output_folder = Path("Cleaned_Data")
output_folder.mkdir(exist_ok=True)

# Store all cleaned datasets
cleaned_datasets = {
    "Customers": customers_clean,
    "Data_Dictionary": dictionary_clean,
    "Exchange_Rates": exchange_clean,
    "Products": products_clean,
    "Sales": sales_clean,
    "Stores": stores_clean
}

# Export each dataset
for name, df in cleaned_datasets.items():

    file_path = output_folder / f"{name}_Cleaned.csv"

    df.to_csv(file_path, index=False)

    print(f"{name} saved successfully!")

print("\nAll six cleaned datasets exported!")
```

```python
# ==========================================
# POSTGRESQL DATABASE CONNECTION
# ==========================================

import psycopg2
from psycopg2 import sql
from sqlalchemy import create_engine, URL
from getpass import getpass

# PostgreSQL credentials
DB_HOST = "localhost"
DB_PORT = "5432"
DB_USER = "postgres"
DB_PASSWORD = getpass("Enter PostgreSQL password: ")

DB_NAME = "global_electronics"

# 1. Connect to default PostgreSQL database
conn = psycopg2.connect(
    host=DB_HOST,
    port=DB_PORT,
    user=DB_USER,
    password=DB_PASSWORD,
    dbname="postgres"
)

# 2. Create database if it does not exist
conn.autocommit = True

cursor = conn.cursor()

cursor.execute(
    "SELECT 1 FROM pg_database WHERE datname = %s",
    (DB_NAME,)
)

database_exists = cursor.fetchone()

if not database_exists:
    cursor.execute(
        sql.SQL("CREATE DATABASE {}").format(
            sql.Identifier(DB_NAME)
        )
    )
    print("Database created successfully!")
else:
    print("Database already exists!")

cursor.close()
conn.close()

# 3. Create SQLAlchemy connection
connection_url = URL.create(
    drivername="postgresql+psycopg2",
    username=DB_USER,
    password=DB_PASSWORD,
    host=DB_HOST,
    port=DB_PORT,
    database=DB_NAME
)

engine = create_engine(connection_url)

# 4. Test connection
with engine.connect() as connection:
    print("Successfully connected to PostgreSQL!")
```

```python
# ==========================================
# UPLOAD ALL SIX DATASETS TO POSTGRESQL
# ==========================================

# Dictionary containing cleaned datasets

cleaned_datasets = {
    "customers": customers_clean,
    "data_dictionary": dictionary_clean,
    "exchange_rates": exchange_clean,
    "products": products_clean,
    "sales": sales_clean,
    "stores": stores_clean
}

# Upload datasets to PostgreSQL

for table_name, df in cleaned_datasets.items():

    print(f"\nUploading {table_name}...")

    df.to_sql(
        name=table_name,
        con=engine,
        if_exists="replace",
        index=False,
        chunksize=1000,
        method="multi"
    )

    print(f"{table_name} uploaded successfully!")
    print("Total records:", len(df))

print("\nAll six datasets uploaded successfully!")
```

```python
# ==========================================
# VERIFY POSTGRESQL TABLES
# ==========================================

from sqlalchemy import inspect, text

inspector = inspect(engine)

tables = inspector.get_table_names()

print("Tables available in PostgreSQL:")
print(tables)

# Check row counts

for table_name in cleaned_datasets.keys():

    query = text(f'SELECT COUNT(*) FROM "{table_name}"')

    with engine.connect() as connection:
        count = connection.execute(query).scalar()

    print(f"{table_name}: {count} records")
```
